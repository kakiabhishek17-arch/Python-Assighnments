```python
1)Sum of Even and Odd Numbers 
Get a number n from the user and calculate the sum of all even numbers and the sum of all odd numbers between 1 and n.
```

```python
es = 0 
os = 0
for i in range(1,21,1):
    if(i%2==0):
        es = es+i
    else:
        os = os+i
print(f"The sum of even numbers:{es}\nThe sum of odd numbers{os}")
```

```python
3)Find the Second-Largest Number Given a list of distinct numbers, 
find the second-largest number without using sort() or max().
```

```python
numbers = [10, 25, 8, 40, 15]
largest = numbers[0]
second = numbers[1]
if second > largest:
    largest, second = second, largest
for num in numbers[2:]:
    if num > largest:
        second = largest
        largest = num
    elif num > second:
        second = num
print("Second-largest number:", second)
```

```python
4)Count Vowels and Consonants Get a string from the user and count the vowels and consonants separately. 
Ignore spaces, digits, and special characters.
```

```python
text = input("Enter a string: ")
v = 0
c = 0
for i in text.lower():
    if i.isalpha():
        if i in "aeiou":
            v += 1
        else:
            c += 1
print("Vowels:", v)
print("Consonants:", c)
```

```python
5)Calculate Factorial Get a positive integer from the user and calculate its factorial using a single for loop. 
Example: 5 → 120
```

```python
n = int(input("enter a number:"))
fact = 1
for i in range(n,0,-1):
    fact = fact*i
print("factorial of this num is:",fact)
```

```python
7)Calculate Electricity Bill Get the units consumed by each of five customers using a list. 
Calculate each customer's bill using these rates:
* First 100 units: ₹2 per unit
* Next 100 units: ₹3 per unit
* Above 200 units: ₹5 per unit
```

```python
unit = []
for i in range(5):
    u = int(input("Enter units consumed: "))
    unit.append(u)
for i in range(5):
    u = unit[i]
    if u <= 100:
        bill = u * 2
    elif u <= 200:
        bill = 100 * 2 + (u - 100) * 3
    else:
        bill = 100 * 2 + 100 * 3 + (u - 200) * 5
    print("Customer", i + 1, "Bill: ₹", bill)
```

```python
8)Reverse a Number Get an integer from the user and reverse its digits using a single for loop. 
Do not convert it into a string. Example: 12345 → 54321
```

```python
n = int(input("Enter an integer: "))
reverse = 0
for i in range(len(str(abs(n)))):
    digit = n % 10
    reverse = reverse * 10 + digit
    n //= 10
print("Reversed number:", reverse)
```

```python
9)Analyze Student Marks Given a list of marks, 
count the students who passed (50 or above), 
failed (below 50), and scored distinction (75 or above)
```

```python
marks = [85, 45, 76, 32, 90, 60, 48, 75]
passed = 0
failed = 0
distinction = 0
for mark in marks:
    if mark >= 50:
        passed += 1
    else:
        failed += 1
    if mark >= 75:
        distinction += 1
print("Passed students:", passed)
print("Failed students:", failed)
print("Students with distinction:", distinction)
```

```python
10)Find Perfect Numbers Print all perfect numbers between 1 and 1000 using a single for loop. 
A perfect number equals the sum of its proper divisors. 
Example: 6 is perfect because 1 + 2 + 3 = 6.
```

```python
for num in range(1, 1001):
    total = 0
    for i in range(1, num):
        if num % i == 0:
            total += i
    if total == num:
        print(num)
```
