```python
1) Leap Year Checker
Accept a year. First check whether it is divisible by 4.
If yes, check whether it is divisible by 100. 
If it is, check whether it is divisible by 400. 
Display whether the year is a leap year.
```

```python
year = float(input("Enter a year:"))
if (year%4==0):
    if(year%100==0):
        if(year%400==0):
            print("Leap year")
        else:
            print("not a leap year")
    else:
        print("Leap year")
else:
    print("not a leap year")
```

```python
2) Triangle Classification
Accept three side lengths. 
First check whether they can form a valid triangle. 
If valid, check whether all three sides are equal, exactly two sides are equal, or all sides are different.
```

```python
a = float(input("Enter first side:"))
b = float(input("Enter second side:"))
c = float(input("Enter third side:"))
if (a+b>c and b+c>a and c+a>b):
    if(a==b and b==c and c==a):
        print("Equivalent triangle")
    elif(a==b or b==c or c==a):
        print("isoceles Triagle")
    else:
        print("scalene triangle")
else:
    print("invalid triangle")
```

```python
4) Password Strength Checker
Accept a password. If its length is at least 8 characters, check whether it contains a digit. 
If it does, check whether it contains an uppercase letter. 
Classify it as strong or moderate based on these conditions. 
Otherwise, classify it as weak.
```

```python
password = input("Enter your password:")
if (len(password)>=8):
    if any(ch.isdigit() for ch in password):
        if any(ch.isupper() for ch in password):
            print("Strong Password")
        else:
            print("Moderate password")
    else:
        print("Moderate password")
else:
    print("Weak password")
```

```python
10) Mini Banking System
Accept account type, account balance, and transaction type (deposit or withdrawal). 
If the account type is savings, check the transaction type. 
For withdrawals, check minimum balance requirements; for deposits, validate the deposit amount. 
If the account type is current, apply a different minimum balance rule.
```

```python
account = input("Enter account type (savings/current): ").lower()
balance = float(input("Enter current balance (₹): "))
transaction = input("Enter transaction (deposit/withdrawal): ").lower()
if account == "savings":
    minimum = 1000
elif account == "current":
    minimum = 5000
else:
    print("Invalid account type")
    minimum = None
if minimum is not None:
    if transaction == "deposit":
        amount = float(input("Enter deposit amount (₹): "))

        if amount > 0:
            balance = balance + amount
            print("Deposit successful")
            print("Updated balance: ₹", balance)
        else:
            print("Invalid deposit amount")
    elif transaction == "withdrawal":
        amount = float(input("Enter withdrawal amount (₹): "))
        if amount > 0:
            if balance - amount >= minimum:
                balance = balance - amount
                print("Withdrawal successful")
                print("Updated balance: ₹", balance)
            else:
                print("Insufficient balance or minimum balance violation")
        else:
            print("Invalid withdrawal amount")
    else:
        print("Invalid transaction type")
```
