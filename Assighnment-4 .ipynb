```python
n = int(input("Enter a number: "))
es = 0
os = 0
i = 1
while i <= n:
    if i % 2 == 0:
        es += i
    else:
        os += i
    i += 1
print("Sum of even numbers:", es)
print("Sum of odd numbers:", os)
```

```python
numbers = [10, 25, 45, 30, 60]
largest = numbers[0]
second = numbers[1]
if second > largest:
    largest, second = second, largest
i = 2
while i < len(numbers):
    if numbers[i] > largest:
        second = largest
        largest = numbers[i]
    elif numbers[i] > second:
        second = numbers[i]
    i += 1
print("Second largest number:", second)
```

```python
text = input("Enter a string: ")
vowels = 0
consonants = 0
i = 0
while i < len(text):
    ch = text[i].lower()
    if ch.isalpha():
        if ch in "aeiou":
            vowels += 1
        else:
            consonants += 1
    i += 1
print("Vowels:", vowels)
print("Consonants:", consonants)
```

```python
units = []
i = 0
while i < 5:
    u = int(input("Enter units consumed: "))
    units.append(u)
    i += 1
i = 0
while i < 5:
    u = units[i]
    if u <= 100:
        bill = u * 2
    elif u <= 200:
        bill = 100 * 2 + (u - 100) * 3
    else:
        bill = 100 * 2 + 100 * 3 + (u - 200) * 5
    print("Customer", i + 1, "Bill: ₹", bill)
    i += 1
```

```python
n = int(input("Enter a number: "))
sign = -1 if n < 0 else 1
n = abs(n)
reverse = 0
while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10
print("Reversed number:", reverse * sign)
```

```python
num = int(input("Enter a number:"))
temp = num
rev = 0
for i in range(num):
    if num > 0:
        digit = num % 10
        rev = rev * 10 + digit
        num = num // 10
if temp == rev:
    print("Palindrome")
else:
    print("Not palindrome")
```
